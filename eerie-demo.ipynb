# eerie — thesis pipeline demo

Illustrate a children's story with the **fine-tuned thesis models**:
`ektvho/bart-cnn` (LSG-BART summarizer) → `ektvho/sd-vist` (Stable Diffusion 2.1,
fine-tuned on VIST). Set the runtime to **GPU** (Runtime → Change runtime type → GPU).

```python
# Setup
!git clone https://github.com/beysa/eerie
%cd eerie
!pip install -q -r requirements.txt
```

```python
from eerie import run_pipeline

story = (
    "Once upon a time, at the edge of a vast frozen sea, there lived a curious "
    "penguin named Pip. While the other penguins huddled together, Pip wondered "
    "what lay beyond the towering blue icebergs. One morning Pip waddled further "
    "than any penguin had dared, and discovered a hidden cove glowing with the "
    "northern lights, where a wise old seal told stories of warm green lands far "
    "to the north."
)
```

```python
# Thesis pipeline: fine-tuned BART summary -> one ektvho/sd-vist image per sentence.
# (The first call downloads the fine-tuned models from the Hugging Face Hub.)
result = run_pipeline(story, summarize=True, seed=42)
print("summary sentences:", result["scenes"])
```

```python
from IPython.display import Image, display
for path in result["panel_paths"]:
    display(Image(path))
```
